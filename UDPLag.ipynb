```python
import pandas as pd

file_path = "/content/UDPLag.csv"

benign_data = []
udp_data = []
webddos_data = []

for chunk in pd.read_csv(file_path, chunksize=50000, low_memory=False):
    chunk.columns = chunk.columns.str.strip()

    benign = chunk[chunk["Label"] == "BENIGN"]
    udp = chunk[chunk["Label"] == "UDP-lag"]
    webddos = chunk[chunk["Label"] == "WebDDoS"]

    if len(benign) > 0:
        benign_data.append(benign)

    if len(udp) > 0:
        udp_data.append(udp)

    if len(webddos) > 0:
        webddos_data.append(webddos)

# Combine
benign_df = pd.concat(benign_data, ignore_index=True)
udp_df = pd.concat(udp_data, ignore_index=True)
webddos_df = pd.concat(webddos_data, ignore_index=True)

# Balance UDP-lag with BENIGN
udp_df = udp_df.sample(
    n=len(benign_df),
    random_state=42
)

# Combine all classes
df = pd.concat(
    [benign_df, udp_df, webddos_df],
    ignore_index=True
)

# Shuffle
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

print("Final dataset shape:", df.shape)
print("\nLabel distribution:")
print(df["Label"].value_counts())
```

# **Clean column names and inspect them**

```python
# Clean column names
df.columns = df.columns.str.strip()

print("Number of columns:", len(df.columns))
print("\nColumns:")
for i, col in enumerate(df.columns):
    print(i, ":", col)
```

# **Remove non-useful columns**

```python
# Columns that are identifiers / text / not suitable for our first model
drop_columns = [
    "Unnamed: 0",
    "Flow ID",
    "Source IP",
    "Destination IP",
    "Timestamp",
    "SimillarHTTP",
    "Label"
]

# Create feature dataset
X = df.drop(columns=drop_columns)

# Create binary target
# BENIGN = 0
# Any attack = 1
y = (df["Label"] != "BENIGN").astype(int)

print("Feature dataset shape:", X.shape)
print("Target shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts())

print("\nNon-numeric columns:")
print(X.select_dtypes(include=["object"]).columns.tolist())
```

# **Check missing and infinite values**

```python
import numpy as np

# Count NaN values
nan_count = X.isna().sum().sum()

# Count infinite values
inf_count = np.isinf(X.to_numpy()).sum()

print("Total NaN values:", nan_count)
print("Total infinite values:", inf_count)
```

# **Clean NaN and infinite values**

```python
import numpy as np

# Replace infinite values with NaN
X = X.replace([np.inf, -np.inf], np.nan)

print("NaN after replacing infinity:", X.isna().sum().sum())

# Fill missing values with the median of each column
X = X.fillna(X.median())

print("NaN after cleaning:", X.isna().sum().sum())
print("Infinite values after cleaning:", np.isinf(X.to_numpy()).sum())
```

# **Train/Test Split**

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data :", X_test.shape)

print("\nTraining labels:")
print(y_train.value_counts())

print("\nTesting labels:")
print(y_test.value_counts())
```

# **Train the Random Forest**

```python
from sklearn.ensemble import RandomForestClassifier

# Create Random Forest model
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

# Train the model
model.fit(X_train, y_train)

print("Random Forest training completed successfully!")
```

# **Test the model**

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Predict on unseen test data
y_pred = model.predict(X_test)

# Calculate metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("===== MODEL PERFORMANCE =====")
print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1-Score : {f1:.4f}")

print("\n===== CLASSIFICATION REPORT =====")
print(classification_report(
    y_test,
    y_pred,
    target_names=["BENIGN", "ATTACK"]
))

print("\n===== CONFUSION MATRIX =====")
print(confusion_matrix(y_test, y_pred))
```

# **Test the baseline model on the NTP dataset**

```python
import pandas as pd
import numpy as np
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

ntp_path = "/content/DrDoS_NTP.csv"

# Use the exact feature order used during training
feature_columns = X_train.columns.tolist()

# Store predictions and true labels
all_true = []
all_pred = []

# Use training-data medians for any missing values
training_medians = X_train.median()

print("Testing existing model on DrDoS_NTP.csv...")
print("The model will NOT be retrained.")

for chunk in pd.read_csv(
    ntp_path,
    chunksize=50000,
    low_memory=False
):
    # Clean column names
    chunk.columns = chunk.columns.str.strip()

    # True labels: BENIGN = 0, everything else = attack = 1
    y_chunk = (chunk["Label"] != "BENIGN").astype(int)

    # Select exactly the same 81 features
    X_chunk = chunk[feature_columns].copy()

    # Convert infinity to NaN
    X_chunk = X_chunk.replace([np.inf, -np.inf], np.nan)

    # Use training medians to fill missing values
    X_chunk = X_chunk.fillna(training_medians)

    # Predict using the EXISTING model
    predictions = model.predict(X_chunk)

    all_true.extend(y_chunk.tolist())
    all_pred.extend(predictions.tolist())

print("\nPrediction completed.")

# Calculate metrics
accuracy = accuracy_score(all_true, all_pred)
precision = precision_score(all_true, all_pred, zero_division=0)
recall = recall_score(all_true, all_pred, zero_division=0)
f1 = f1_score(all_true, all_pred, zero_division=0)

print("\n===== NTP UNSEEN-DATA PERFORMANCE =====")
print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1-Score : {f1:.4f}")

print("\n===== CONFUSION MATRIX =====")
print(confusion_matrix(all_true, all_pred))

print("\nTotal NTP samples tested:", len(all_true))
```
